### Popularity based recommendation system

```python
ratings_with_book_name=ratings.merge(books, on='ISBN')
```

```python
ratings_with_book_name
```

```python
number_of_rating_df=ratings_with_book_name.groupby('Book-Title')['Book-Rating'].count().reset_index()
number_of_rating_df=number_of_rating_df.rename(columns={'Book-Rating':'No_of_rating'})
number_of_rating_df
```

```python
average_rating_df=ratings_with_book_name.groupby('Book-Title')['Book-Rating'].mean().reset_index()
average_rating_df=average_rating_df.rename(columns={'Book-Rating':'Average_rating'})
average_rating_df
```

```python
popularity_df=number_of_rating_df.merge(average_rating_df,on='Book-Title')
popularity_df
```

```python
Top_50_books=popularity_df[popularity_df['No_of_rating']>=250].sort_values('Average_rating',ascending=False).head(50)
Top_50_books
```

```python
Top_50_books=Top_50_books.merge(books,on='Book-Title')
Top_50_books
```

```python
# harry potter's same book has different isbn so we got 196 rows here. so  dropping the duplicate
```

```python
Top_50_books=Top_50_books.drop_duplicates('Book-Title')[['Book-Title','Book-Author','Year-Of-Publication','Image-URL-M','No_of_rating','Average_rating']]
Top_50_books.shape
```

### Collaborative filterting based recommender

```python
active_rater=(ratings_with_book_name.groupby('User-ID').count()['Book-Rating'] > 200)
active_rater=active_rater[active_rater].index
active_rater
```

```python
filtered_rating_with_book_name=ratings_with_book_name[ratings_with_book_name['User-ID'].isin(active_rater)]
filtered_rating_with_book_name
```

```python
enough_rated_books=filtered_rating_with_book_name.groupby('Book-Title')['Book-Rating'].count() >=50
famous_boooks=enough_rated_books[enough_rated_books].index
famous_boooks
```

```python
final_rating=filtered_rating_with_book_name[filtered_rating_with_book_name['Book-Title'].isin(famous_boooks)]
final_rating
```

```python
pt=final_rating.pivot_table(index='Book-Title',columns='User-ID',values='Book-Rating')
pt
```

```python
pt.fillna(0,inplace=True)
pt
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity_score=cosine_similarity(pt)
```

```python
similarity_score[0]
```

```python
book_index=np.where(pt.index=='1984')[0][0]
sorted(list(enumerate(similarity_score[book_index])),reverse=True,key=lambda x:x[1])[1:6]
```

```python
def recommend(book_name):
    #index fetch 
    book_index=np.where(pt.index==book_name)[0][0]

    similar_items=sorted(list(enumerate(similarity_score[book_index])),reverse=True,key=lambda x:x[1])[1:6]

    for i in similar_items:
        # recommendation=pt[similar_items[0]]
        print(pt.index[i[0]])

    # return recommendation
```

```python
recommend('Harry Potter and the Chamber of Secrets (Book 2)')
```
